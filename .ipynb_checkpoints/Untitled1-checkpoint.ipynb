```python
import pandas as pd

d = {
    "Name":["Aman","Amit","Kunal"],
    "Age":[22,33,44],
    "Course":["python","java","c++"]
}

df = pd.DataFrame(d)

print(df.loc[1 , "Age"])
```

```python
# groupby()


d = {
    "Product":["1+","Vivo","1+","Nokia","1+","Vivo"],
    "Order":[2,1,3,4,1,1]
}

df = pd.DataFrame(d)

group = df.groupby(['Product'])

result = group.sum()


print(result)
```
