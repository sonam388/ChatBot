```python
import pandas as pd
import numpy as np 

arr = np.array([1,-2,3,-4,0])

a = np.count_nonzero(arr > 0)
b = np.count_nonzero(arr < 0)
c = np.count_nonzero(arr == 0)

print(a)
print(b)
print(c)
```

```python
import numpy as np

arr = np.array([10,20,5,30,35,40,5,60])


result = np.where(
    arr < 10 , 0  , np.where(
        arr < 50 , 1, 2
    )
)
print(result)
```

```python
import numpy as np

arr = np.array([1,2,1,3,4,2,4,5,5])

for x in arr:
    if np.count_nonzero(arr == x) > 1:
        print(x)
```
