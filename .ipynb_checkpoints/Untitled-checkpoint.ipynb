```python
import json 
import random
import joblib

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Load DataSet ------>

with open ("data.json" , "r") as file:
    data = json.load(file)

sentence = []
label = []
responce = {}

for i in data['intents']:
    tag = i['tag']

    responce[tag] = i["patterns"]

    for patten in i['patterns']:
        sentence.append(patten)
        label.append(tag)
        
print(sentence)
print(label)

# Train Test Split ----->



x_train , x_test , y_train , y_test = train_test_split(sentence , label )

# TF-TDF  ------------->

vector = TfidfVectorizer()

x_train_n = vector.fit_transform(x_train).toarray()
x_test_n = vector.transform(x_test).toarray()

#  Train Model ----> 

model = GaussianNB()
model.fit(x_train_n , y_train)


# Test Model --->

y_predict = model.predict(x_test_n)

score = accuracy_score(y_test ,y_predict )

print("Model Acuracy " ,score)

# Save Model ---> 

joblib.dump(model , "Chatbot_model.pkl")
joblib.dump(vector , "tfidf.pkl")
joblib.dump(responce , "responses.pkl")

```
